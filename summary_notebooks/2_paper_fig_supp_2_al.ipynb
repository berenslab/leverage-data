```python
project_dir = '.'

CLEANED_FILE = 'areds_cleaned.csv'
metadata_dir = os.path.join(project_dir, 'metadata')
log_dir = 'log_dir'

CHKPT_DIR = chkpt_dir = os.path.join(project_dir, 'chkpt_dir')
RESULTS_DIR = os.path.join(project_dir, 'results')
stylef = './mpl.mplstyle'
img_dir = f"./images"
img_dir
```

```python
or_df = pd.read_csv(os.path.join(log_dir, f'{CLEANED_FILE}'))
print(or_df.shape)
or_df.head(2)
```

```python
or_df['summary'].value_counts()
```

```python
metadata_df = pd.read_csv(os.path.join(metadata_dir,'surv_metadata.csv'))
metadata_df.head()

patient_identifier= 'patient_id'
SPLIT_ID = 11042026
split_ids_file = os.path.join(metadata_dir, f"splits_patient-ids-{SPLIT_ID}.pkl")
dict_seed_splits_patientids = pickle.load(open(split_ids_file, "rb"))

test = metadata_df[metadata_df[patient_identifier].isin(dict_seed_splits_patientids[SPLIT_ID]["test"])].reset_index(drop=True)
test.head()
```

```python
from plotting_utils import plot_one_image_curve, draw_patient_grid, get_patient_panel
```

```python
CHKPT_DIR = 'curves_to_plot'
stylefile = 'mpl.mplstyle'
figures_dir = 'figures/main'
survival_head = 'mlp' 
```

```python
all_pid = plot_one_image_curve(or_df, test, freeze_encoder = 1,
                                train_size = 500, has_event=True, 
                                CHKPT_DIR=CHKPT_DIR, return_only_pid = True,
                                survival_head = survival_head
                                        )
```

```python
cleaned_pid = []
for i in range(len(all_pid)):
    pid = all_pid[i][0]
    cleaned_pid.append(pid)
len(cleaned_pid)
```

```python
for i, pid in enumerate(cleaned_pid[7:8]):
    records = []

    t1 = get_patient_panel(or_df, test, 0, freeze_encoder = 1,
                                       train_size = 1000, has_event=True, CHKPT_DIR=CHKPT_DIR, 
                                       patient_id = pid, survival_head=survival_head )
    records.append(t1)
    t2 = get_patient_panel(or_df, test, 0, freeze_encoder = 1,
                                           train_size = 32250, has_event=True, CHKPT_DIR=CHKPT_DIR, 
                                           patient_id = pid, survival_head=survival_head )
    records.append(t2)
    t3 = get_patient_panel(or_df, test, 0, freeze_encoder = 0,
                                           train_size = 1000, has_event=True, CHKPT_DIR=CHKPT_DIR, 
                                           patient_id = pid, survival_head=survival_head )
    records.append(t3)
    t4 = get_patient_panel(or_df, test, 0, freeze_encoder = 0,
                                           train_size = 32250, has_event=True, CHKPT_DIR=CHKPT_DIR, 
                                           patient_id = pid, survival_head=survival_head )
    records.append(t4)

    draw_patient_grid(records, image_dir=img_dir, 
                        n_cols=4,
                        legend_bbox_to_anchor=(0.6, 0.27),  
                        text_x =0.33,
                        text_y=1.14,
                        save_path=f'{figures_dir}',
                        stylef = stylef,
                        save_name = f"{pid}_{i}_{survival_head}",
                        show_title = False,
                        survival_head=survival_head)    

    
```
